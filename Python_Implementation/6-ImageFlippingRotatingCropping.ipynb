```python
# Import for OpenCV
import cv2

import numpy as np
import matplotlib.pyplot as plt 
```

```python

# Load the image and get it as BGR
image = cv2.imread("./images/giraf.jpg")
print(image)
```

```python
# convert to RGB image
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image_rgb)
```

Flip the Image


```python
# Choose 1  flip image Horizontally 
# Choose 0  flip image Vertically 
# Choose -1 flip image on both axes  
flip_horizontal = cv2.flip(image, 1)
flip_vertical   = cv2.flip(image, 0) 
flip_both       = cv2.flip(image, -1)
```

```python
# Plot
titles = [ "Original image" , "Horizontal Flip", "Vertical Flip", "Flip Both axes" ]
images = [image_rgb,
                    cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB),
                    cv2.cvtColor(flip_vertical, cv2.COLOR_BGR2RGB),
                    cv2.cvtColor(flip_both, cv2.COLOR_BGR2RGB)]

plt.figure(figsize=(12,12))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## To Rotate image 

```python
# Rotate methods convert from BGR to RGB
rotated_image_90_cw = cv2.rotate(image_rgb, cv2.ROTATE_90_CLOCKWISE)
rotated_image_90_ccw = cv2.rotate(image_rgb, cv2.ROTATE_90_COUNTERCLOCKWISE)
rotated_image_180 = cv2.rotate(image_rgb, cv2.ROTATE_180)
```

```python
# Plot rotated images
# Plot
titles = [ "Original image" , "90 ClockWise", "90 Counter ClockWise", "180 Rotate" ]
images = [image_rgb,
                    rotated_image_90_cw,
                    rotated_image_90_ccw,
                    rotated_image_180]

plt.figure(figsize=(12,12))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## Cropping an image

```python
# Getting the dimension or shape of the image
print(image_rgb.shape)
```

```python
# We need to have x,y coordinates for start and end 
x_start, y_start = 200, 400
x_end , y_end = 800, 1000

cropped_image = image_rgb[y_start:y_end, x_start:x_end]
```

```python
# Show all resized images with each algorithm
plt.figure(figsize=(12,8))

# Show the original Image
plt.subplot(1,2,1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

# Show cropped image
plt.subplot(1,2,2)
plt.imshow(cropped_image)
plt.title("Cropped Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
